### Task 1: Load and Inspect the Dataset

Loads the original Chicago crime dataset into a Pandas DataFrame using line-by-line CSV parsing to handle malformed quotation marks. Displays the first 10 records, schema, data types and validates the dataset dimensions of 498 rows and 20 columns.

```python

# ============================================================
# USE CASE 1 - CODE CELL 1
# TASK 1: LOAD THE DATASET INTO A PANDAS DATAFRAME
# ============================================================

# Assessment Requirements:
# 1. Load the Chicago crime CSV dataset.
# 2. Display the first 10 rows.
# 3. Show the schema and column data types.
# 4. Find the total number of rows and columns.
#
# Methods Used:
# Path() locates the project and CSV dataset.
# csv.reader() handles CSV quotation marks.
# pd.DataFrame() loads records into Pandas.
# head() displays the first 10 rows.
# info() displays the schema and data types.
# shape calculates the total rows and columns.
#
# Source Data Note:
# The original crime CSV has malformed quotation marks
# in its final two location fields.
# Reading each physical CSV line independently preserves
# all records without changing or deleting source data.

# STEP 1: Import the required libraries.
from pathlib import Path
import csv
import pandas as pd
from IPython.display import display

# STEP 2: Identify the new project folder.
BASE_DIR = Path.cwd().resolve()

# Jupyter may run inside the notebooks directory.
if BASE_DIR.name == "notebooks":
    BASE_DIR = BASE_DIR.parent


assert BASE_DIR.name == "Chicago_Crime_Capstone_2026", (
    f"Incorrect project folder: {BASE_DIR}"
)

# STEP 3: Locate the original crime CSV.
CRIME_FILE = (
    BASE_DIR / "data" / "raw" /
    "chicago_crime_dataset.csv"
)

assert CRIME_FILE.is_file(), (
    f"Crime dataset not found: {CRIME_FILE}"
)

# STEP 4: Read the source records.
# splitlines() works here because the uploaded source
# contains one crime record per physical line.
with CRIME_FILE.open(
    mode="r",
    encoding="utf-8-sig",
    newline=""
) as file:
    raw_lines = [
        line
        for line in file.read().splitlines()
        if line.strip()
    ]

# STEP 5: Parse individual CSV lines.
# This avoids the malformed quotation marks joining
# the final two crime records during normal CSV loading.
parsed_rows = [
    next(csv.reader([line]))
    for line in raw_lines
]

# STEP 6: Separate header and crime records.
column_names = parsed_rows[0]
crime_records = parsed_rows[1:]

# STEP 7: Validate the original CSV structure.
# Every record must contain the same number of fields.
expected_columns = len(column_names)

for line_number, row in enumerate(
    crime_records,
    start=2
):
    assert len(row) == expected_columns, (
        f"Line {line_number}: "
        f"Expected {expected_columns} fields, "
        f"found {len(row)}"
    )

# STEP 8: Load the records into a Pandas DataFrame.
# Keep source values as text for initial inspection.
# Cleaning and type conversions happen in Task 2.
crime_df = pd.DataFrame(
    crime_records,
    columns=column_names
)

# STEP 9: Display the first 10 rows.
print("========== FIRST 10 CRIME RECORDS ==========")

display(crime_df.head(10))

# STEP 10: Display the dataset schema.
print("\n========== DATASET SCHEMA ==========")

crime_df.info()

# STEP 11: Display column data types.
print("\n========== COLUMN DATA TYPES ==========")

print(crime_df.dtypes)

# STEP 12: Find the total rows and columns.
total_rows, total_columns = crime_df.shape

print("\n========== DATASET DIMENSIONS ==========")

print("Total Rows:", total_rows)
print("Total Columns:", total_columns)

# STEP 13: Validate against the uploaded source.
assert total_rows == 498
assert total_columns == 20

# Confirm no source crime record was discarded.
assert total_rows == len(crime_records)

# STEP 14: Answer the assessment question.
print("\n========== ASSESSMENT ANSWER ==========")

print("Question: How many rows and columns are there?")

print(
    f"Answer: The Chicago crime dataset contains "
    f"{total_rows} rows and {total_columns} columns."
)

print("\nUSE CASE 1 - TASK 1: PASS")
```

### Task 2: Clean the Chicago Crime Dataset

Cleans the dataset by converting dates to datetime format, handling missing values, removing extra whitespace and standardizing categorical fields. Validates data integrity and saves the cleaned dataset without removing any original records or columns.

```python

# ============================================================
# USE CASE 1 - CODE CELL 2
# TASK 2: CLEAN THE CHICAGO CRIME DATASET
# ============================================================

# Assessment Requirements:
# 1. Convert Date into datetime format.
# 2. Identify and handle missing values for key fields.
# 3. Standardize categorical fields:
#    - Remove extra whitespace.
#    - Use a consistent text case.
#
# Methods Used:
# copy() preserves the original loaded DataFrame.
# str.strip() removes leading and trailing spaces.
# replace() converts empty strings into missing values.
# isna().sum() counts missing values.
# pd.to_datetime() converts date strings to datetime.
# fillna() handles missing location descriptions.
# str.upper() standardizes categorical text.
# to_csv() saves the cleaned dataset.
#
# Important:
# No crime records or original columns are discarded.
# Missing geographic coordinates remain missing
# because inventing coordinates would be inaccurate.


# ============================================================
# STEP 1: CREATE A COPY OF THE ORIGINAL DATASET
# ============================================================

# Keep the DataFrame from Code Cell 1 unchanged.
clean_df = crime_df.copy()

original_rows = len(clean_df)

# Standardize column names for consistent processing.
clean_df.columns = (
    clean_df.columns.str.strip().str.lower()
)


# ============================================================
# STEP 2: REMOVE EXTRA WHITESPACE
# ============================================================

# Process all source columns as text before converting
# the required date column into its correct data type.
for column in clean_df.columns:

    clean_df[column] = (
        clean_df[column]
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
    )

# Empty strings are now represented as missing values.


# ============================================================
# STEP 3: IDENTIFY MISSING VALUES IN KEY FIELDS
# ============================================================

# These fields are important for crime identification,
# classification, geographic analysis and arrest analysis.
key_fields = [
    "id",
    "case_number",
    "date",
    "iucr_code",
    "location_desc",
    "arrest",
    "domestic",
    "district_code",
    "community_code"
]

missing_before = (
    clean_df[key_fields]
    .isna()
    .sum()
)

print("========== MISSING VALUES BEFORE CLEANING ==========")

display(
    missing_before
    .reset_index()
    .rename(columns={
        "index": "column_name",
        0: "missing_count"
    })
)


# ============================================================
# STEP 4: CONVERT DATE INTO DATETIME FORMAT
# ============================================================

# The uploaded crime CSV uses month/day/year dates.
# errors="coerce" converts invalid dates into NaT.
clean_df["date"] = pd.to_datetime(
    clean_df["date"],
    format="%m/%d/%Y",
    errors="coerce"
)

# Confirm that every crime date was converted.
invalid_dates = clean_df["date"].isna().sum()

print("\n========== DATE CONVERSION ==========")

print("Date Data Type:", clean_df["date"].dtype)
print("Invalid or Missing Dates:", invalid_dates)

assert invalid_dates == 0, (
    "Some crime dates could not be converted."
)


# ============================================================
# STEP 5: HANDLE MISSING KEY FIELD VALUES
# ============================================================

# Missing location descriptions are labeled UNKNOWN.
# This keeps the corresponding crime records available
# for analysis without inventing a specific location.
clean_df["location_desc"] = (
    clean_df["location_desc"].fillna("UNKNOWN")
)

# Essential identifiers and classification fields must
# remain present. We do not invent missing identifiers.
essential_fields = [
    "id",
    "case_number",
    "date",
    "iucr_code",
    "arrest",
    "domestic",
    "district_code",
    "community_code"
]

assert clean_df[essential_fields].notna().all().all(), (
    "Missing essential crime field values detected."
)

# Missing coordinates are preserved as missing values.
# Their actual geographic positions are unknown.


# ============================================================
# STEP 6: STANDARDIZE CATEGORICAL FIELDS
# ============================================================

# Convert text categories to uppercase for consistency.
categorical_fields = [
    "case_number",
    "block",
    "iucr_code",
    "location_desc",
    "fbi_code",
    "arrest",
    "domestic"
]

for column in categorical_fields:

    clean_df[column] = (
        clean_df[column]
        .str.upper()
    )

# Check the valid Boolean categories.
assert set(clean_df["arrest"].dropna().unique()) <= {
    "TRUE", "FALSE"
}

assert set(clean_df["domestic"].dropna().unique()) <= {
    "TRUE", "FALSE"
}


# ============================================================
# STEP 7: CHECK MISSING VALUES AFTER CLEANING
# ============================================================

missing_after = (
    clean_df[key_fields]
    .isna()
    .sum()
)

print("\n========== MISSING VALUES AFTER CLEANING ==========")

display(
    missing_after
    .reset_index()
    .rename(columns={
        "index": "column_name",
        0: "missing_count"
    })
)


# ============================================================
# STEP 8: VERIFY DATA PRESERVATION
# ============================================================

# Ensure no source records or columns were removed.
assert len(clean_df) == original_rows
assert clean_df.shape[1] == crime_df.shape[1]

# Check that crime identifiers remain unique.
assert clean_df["id"].is_unique

print("\n========== DATA PRESERVATION ==========")

print("Original Records:", original_rows)
print("Cleaned Records:", len(clean_df))
print("Records Discarded:", original_rows - len(clean_df))

print("Cleaned Columns:", clean_df.shape[1])


# ============================================================
# STEP 9: SAVE THE CLEANED DATASET
# ============================================================

# Save the intermediate cleaning output.
# This file will be used to generate date features
# in the next assessment task.
CLEANED_DIR = BASE_DIR / "data" / "cleaned"

CLEANED_DIR.mkdir(
    parents=True,
    exist_ok=True
)

CLEANED_FILE = (
    CLEANED_DIR / "chicago_crime_cleaned.csv"
)

clean_df.to_csv(
    CLEANED_FILE,
    index=False
)

assert CLEANED_FILE.is_file()

print("\nCleaned Dataset Saved:", CLEANED_FILE.name)


# ============================================================
# STEP 10: DISPLAY CLEANED DATA
# ============================================================

print("\n========== FIRST 10 CLEANED RECORDS ==========")

display(clean_df.head(10))


# ============================================================
# STEP 11: FINAL VALIDATION
# ============================================================

assert len(clean_df) == 498
assert clean_df.shape[1] == 20
assert clean_df["date"].notna().all()
assert clean_df["location_desc"].notna().all()

print("\nUSE CASE 1 - TASK 2: PASS")
```

### Task 3: Generate New Features

Extracts Year, Month and DayOfWeek from the date column using Pandas datetime methods. Validates the existing year values and generated features, then saves the feature-engineered dataset with 22 columns while preserving all original records.

```python

# ============================================================
# USE CASE 1 - CODE CELL 3
# TASK 3: GENERATE NEW FEATURES
# ============================================================

# Assessment Requirement:
# Extract Year, Month, and DayOfWeek from the Date field.
#
# Required Methods:
# dt.year extracts the year from a datetime column.
# dt.month extracts the month number (1 to 12).
# dt.day_name() extracts the weekday name.
#
# Additional Methods:
# copy() preserves the previously cleaned DataFrame.
# to_numeric() checks the original Year values.
# to_csv() saves the feature-engineered dataset.
#
# Important:
# The original dataset already contains Year.
# We compare it with the year extracted from Date
# before updating the existing Year column.
# No crime records or original fields are discarded.


# ============================================================
# STEP 1: COPY THE CLEANED DATASET
# ============================================================

# Use the cleaned DataFrame created in Code Cell 2.
feature_df = clean_df.copy()

original_rows = len(feature_df)

# Verify that Date has already been converted.
assert pd.api.types.is_datetime64_any_dtype(
    feature_df["date"]
), "Date must be converted to datetime before this task."


# ============================================================
# STEP 2: CHECK THE EXISTING YEAR COLUMN
# ============================================================

# Convert the original Year values into numbers.
original_year = pd.to_numeric(
    feature_df["year"],
    errors="coerce"
)

# Extract the year directly from the Date column.
extracted_year = feature_df["date"].dt.year

# Count differences between the original Year and
# the year calculated from the actual crime date.
year_mismatches = (
    original_year.ne(extracted_year)
    | original_year.isna()
).sum()

print("========== EXISTING YEAR VALIDATION ==========")

print("Year Mismatches:", int(year_mismatches))


# ============================================================
# STEP 3: EXTRACT YEAR FROM DATE
# ============================================================

# dt.year returns the calendar year of each crime date.
# Update the existing Year column rather than creating
# an unnecessary duplicate column.
feature_df["year"] = feature_df["date"].dt.year


# ============================================================
# STEP 4: EXTRACT MONTH FROM DATE
# ============================================================

# dt.month returns numbers from 1 to 12.
# Example: May becomes 5.
feature_df["month"] = feature_df["date"].dt.month


# ============================================================
# STEP 5: EXTRACT DAY OF WEEK FROM DATE
# ============================================================

# dt.day_name() returns weekday names.
# Example: Monday, Tuesday, Wednesday.
feature_df["day_of_week"] = (
    feature_df["date"].dt.day_name()
)


# ============================================================
# STEP 6: DISPLAY THE GENERATED FEATURES
# ============================================================

print("\n========== GENERATED DATE FEATURES ==========")

display(
    feature_df[
        ["date", "year", "month", "day_of_week"]
    ].head(10)
)


# ============================================================
# STEP 7: CHECK FEATURE DATA TYPES
# ============================================================

print("\n========== FEATURE DATA TYPES ==========")

print(
    feature_df[
        ["date", "year", "month", "day_of_week"]
    ].dtypes
)


# ============================================================
# STEP 8: VALIDATE THE GENERATED FEATURES
# ============================================================

# Check that Year, Month, and DayOfWeek exist.
required_features = [
    "year",
    "month",
    "day_of_week"
]

assert all(
    column in feature_df.columns
    for column in required_features
)

# Verify that every Month is between 1 and 12.
assert feature_df["month"].between(1, 12).all()

# Verify that every weekday is valid.
valid_weekdays = {
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
}

assert set(
    feature_df["day_of_week"].dropna().unique()
).issubset(valid_weekdays)

# Confirm that no generated feature is missing.
assert feature_df[required_features].notna().all().all()

# Confirm the number of records is unchanged.
assert len(feature_df) == original_rows

# Original CSV: 20 columns.
# Month and DayOfWeek add two new columns.
# Year already exists and is updated.
assert feature_df.shape[1] == 22


# ============================================================
# STEP 9: SAVE THE FEATURE-ENGINEERED DATASET
# ============================================================

# This dataset will be used by Notebooks 02 and 03.
FEATURES_FILE = (
    CLEANED_DIR / "chicago_crime_features.csv"
)

feature_df.to_csv(
    FEATURES_FILE,
    index=False
)

# Confirm the file was generated successfully.
assert FEATURES_FILE.is_file()


# ============================================================
# STEP 10: DISPLAY THE FINAL TASK SUMMARY
# ============================================================

print("\n========== FEATURE ENGINEERING SUMMARY ==========")

print("Total Crime Records:", len(feature_df))
print("Total Columns:", feature_df.shape[1])

print("Year Extracted: YES")
print("Month Extracted: YES")
print("DayOfWeek Extracted: YES")

print("Records Discarded:", original_rows - len(feature_df))

print("Saved Dataset:", FEATURES_FILE.name)

print("\nUSE CASE 1 - TASK 3: PASS")
```

### Task 4: Missing Value Analysis Using NumPy

Uses NumPy to calculate missing-value counts and percentages for each column and identifies columns exceeding the 50% missing-value threshold. Evaluates whether columns should be dropped and validates the calculations against Pandas without modifying the dataset.

```python

# ============================================================
# USE CASE 1 - CODE CELL 4
# TASK 4: MISSING VALUE ANALYSIS USING NUMPY
# ============================================================

# Assessment Requirements:
# 1. Calculate the percentage of missing values per column.
# 2. Identify columns with more than 50% missing values.
# 3. Answer whether any columns should be dropped.
#
# Methods Used:
# isna() identifies missing values.
# to_numpy() converts missing indicators to a NumPy array.
# np.sum() calculates missing counts.
# np.mean() calculates missing percentages.
# np.round() rounds percentages to two decimal places.



# ============================================================
# STEP 1: IMPORT REQUIRED LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
from IPython.display import display


# ============================================================
# STEP 2: VALIDATE THE FEATURE-ENGINEERED DATASET
# ============================================================

# feature_df was created in Code Cell 3.
assert "feature_df" in globals(), (
    "Run Code Cell 3 before Code Cell 4."
)

# Work with a copy to preserve the existing DataFrame.
missing_df = feature_df.copy()

total_rows, total_columns = missing_df.shape

print("========== DATASET INFORMATION ==========")
print("Total Rows:", total_rows)
print("Total Columns:", total_columns)

# Verify the expected dataset dimensions.
assert total_rows == 498, (
    f"Expected 498 rows, found {total_rows}"
)

assert total_columns == 22, (
    f"Expected 22 columns, found {total_columns}"
)


# ============================================================
# STEP 3: IDENTIFY MISSING VALUES USING NUMPY
# ============================================================

# Convert missing-value indicators into a NumPy array.
# True means missing; False means present.
missing_array = missing_df.isna().to_numpy(dtype=bool)

# Count missing values for every column.
# axis=0 performs the calculation column-wise.
missing_counts = np.sum(
    missing_array,
    axis=0
)

# Calculate missing-value percentages.
# np.mean() finds the fraction of missing entries.
missing_percentages = (
    np.mean(missing_array, axis=0) * 100
)


# ============================================================
# STEP 4: DISPLAY MISSING VALUES FOR EACH COLUMN
# ============================================================

# Create a summary table with counts and percentages.
missing_summary = pd.DataFrame({
    "column_name": missing_df.columns,
    "missing_count": missing_counts.astype(int),
    "missing_percentage": np.round(
        missing_percentages,
        2
    )
})

print("\n========== MISSING VALUES PER COLUMN ==========")

display(missing_summary)


# ============================================================
# STEP 5: IDENTIFY COLUMNS ABOVE 50% MISSING
# ============================================================

# The assessment specifies more than 50%.
# Therefore, use > 50 rather than >= 50.
high_missing_columns = missing_summary.loc[
    missing_summary["missing_percentage"] > 50
].copy()

print("\n========== COLUMNS ABOVE 50% MISSING ==========")

if high_missing_columns.empty:
    print("No columns have more than 50% missing values.")
else:
    display(high_missing_columns)


# ============================================================
# STEP 6: ANSWER THE ASSESSMENT QUESTION
# ============================================================

print("\n========== ASSESSMENT QUESTION AND ANSWER ==========")

print(
    "Question: Can you drop any columns with "
    "more than 50% missing values?"
)

if high_missing_columns.empty:

    print(
        "Answer: No. None of the columns contains "
        "more than 50% missing values."
    )

    print(
        "No columns need to be dropped based "
        "on the specified threshold."
    )

else:

    print(
        "Answer: Yes. The following columns exceed "
        "the 50% missing-value threshold:"
    )

    print(
        high_missing_columns["column_name"].tolist()
    )

    print(
        "These columns are candidates for removal, "
        "but their analytical importance should be "
        "evaluated before dropping them."
    )


# ============================================================
# STEP 7: DISPLAY THE MISSING-VALUE SUMMARY
# ============================================================

# Count columns containing at least one missing value.
columns_with_missing = int(
    np.sum(missing_counts > 0)
)

# Find the largest missing-value percentage.
maximum_missing = float(
    np.max(missing_percentages)
)

print("\n========== MISSING-VALUE SUMMARY ==========")

print("Total Crime Records:", total_rows)
print("Total Columns:", total_columns)

print(
    "Columns With Missing Values:",
    columns_with_missing
)

print(
    "Maximum Missing Percentage:",
    round(maximum_missing, 2),
    "%"
)

print(
    "Columns Above 50% Missing:",
    len(high_missing_columns)
)


# ============================================================
# STEP 8: VALIDATE NUMPY CALCULATIONS
# ============================================================

# Confirm NumPy missing counts match Pandas.
assert np.array_equal(
    missing_counts,
    missing_df.isna().sum().to_numpy()
), "NumPy and Pandas missing counts do not match."

# All percentages must lie between 0 and 100.
assert np.all(
    (missing_percentages >= 0)
    &
    (missing_percentages <= 100)
), "Invalid missing-value percentage found."

# Confirm every missing count is valid.
assert np.all(
    missing_counts <= total_rows
)

# Confirm original data was not modified or discarded.
assert feature_df.shape == (498, 22)

print("\nUSE CASE 1 - TASK 4: PASS")
```

### Task 5: Design and Populate the SQLite Database

Uses SQLAlchemy to design the crime table schema, convert fields into appropriate data types and insert the cleaned crime records into SQLite. Validates the primary key, record count, column count and database integrity while preventing duplicate insertion.

```python

# ============================================================
# USE CASE 1 - CODE CELL 5
# TASK 5: DESIGN AND POPULATE THE SQLITE DATABASE
# ============================================================

# Assessment Requirements:
# 1. Design the crime table schema.
# 2. Insert the cleaned crime data using SQLAlchemy.
#
# Methods Used:
# create_engine() connects Python to SQLite3.
# Table() and Column() define the database schema.
# create_all() creates the table if it does not exist.
# pd.to_numeric() converts numeric fields.
# to_sql() inserts Pandas records into the SQL table.
# SQL SELECT queries validate the inserted records.



# ============================================================
# STEP 1: IMPORT DATABASE LIBRARIES
# ============================================================

import pandas as pd
from pathlib import Path
from IPython.display import display

from sqlalchemy import (
    create_engine,
    MetaData,
    Table,
    Column,
    Integer,
    Float,
    String,
    Text,
    Date,
    Boolean,
    inspect,
    text
)


# ============================================================
# STEP 2: DEFINE THE NEW DATABASE FILE
# ============================================================

# Store the database inside the new rebuild project.
DB_FILE = BASE_DIR / "data" / "chicago_crime.db"

# Connect using SQLAlchemy 2.
engine = create_engine(
    f"sqlite:///{DB_FILE.as_posix()}",
    future=True
)

print("========== DATABASE CONNECTION ==========")
print("Database:", DB_FILE.name)


# ============================================================
# STEP 3: PREPARE THE CLEANED DATA FOR SQL
# ============================================================

# Use the feature-engineered dataset from Code Cell 3.
# Create a copy to avoid changing previous calculations.
db_df = feature_df.copy()

# Convert numeric columns to appropriate numeric types.
integer_columns = [
    "id",
    "beat_num",
    "district_code",
    "ward_no",
    "community_code",
    "x_coordinate",
    "y_coordinate",
    "year",
    "month"
]

for column in integer_columns:

    db_df[column] = pd.to_numeric(
        db_df[column],
        errors="coerce"
    ).astype("Int64")

# Convert geographical coordinates to numeric values.
# Missing coordinates remain NULL in the database.
for column in ["latitude", "longitude"]:

    db_df[column] = pd.to_numeric(
        db_df[column],
        errors="coerce"
    )

# Convert TRUE/FALSE values into database Booleans.
boolean_mapping = {
    "TRUE": True,
    "FALSE": False,
    "1": True,
    "0": False
}

for column in ["arrest", "domestic"]:

    db_df[column] = (
        db_df[column]
        .astype("string")
        .str.strip()
        .str.upper()
        .map(boolean_mapping)
        .astype("boolean")
    )

# Preserve date-only information from the source.
# The dataset does not contain occurrence times.
db_df["date"] = pd.to_datetime(
    db_df["date"],
    errors="coerce"
).dt.date

db_df["date_of_update"] = pd.to_datetime(
    db_df["date_of_update"],
    format="%m/%d/%Y",
    errors="coerce"
).dt.date

# Standardize IUCR codes to four characters.
# Example: 486 becomes 0486.
# This supports matching with the IUCR reference table.
db_df["iucr_code"] = (
    db_df["iucr_code"]
    .astype("string")
    .str.strip()
    .str.upper()
    .str.zfill(4)
)

# Ensure essential crime fields have valid values.
required_fields = [
    "id",
    "case_number",
    "date",
    "iucr_code",
    "arrest",
    "domestic",
    "year",
    "month",
    "day_of_week"
]

assert db_df[required_fields].notna().all().all(), (
    "Missing values found in required database fields."
)

# Verify primary-key and unique-ID requirements.
assert db_df["case_number"].is_unique
assert db_df["id"].is_unique

assert len(db_df) == 498
assert db_df.shape[1] == 22


# ============================================================
# STEP 4: DESIGN THE CRIME TABLE SCHEMA
# ============================================================

# SQLAlchemy MetaData stores the table definition.
metadata = MetaData()

# case_number is the primary key in the assessment
# data model. id is retained as a unique identifier.
crime_table = Table(
    "crime_records",
    metadata,

    Column("id", Integer, unique=True, nullable=False),
    Column("case_number", String(30), primary_key=True),

    Column("date", Date, nullable=False),
    Column("block", Text),
    Column("iucr_code", String(10), nullable=False),
    Column("location_desc", Text),

    Column("arrest", Boolean, nullable=False),
    Column("domestic", Boolean, nullable=False),

    Column("beat_num", Integer),
    Column("district_code", Integer),
    Column("ward_no", Integer),
    Column("community_code", Integer),

    Column("fbi_code", String(10)),

    Column("x_coordinate", Integer),
    Column("y_coordinate", Integer),

    Column("year", Integer),
    Column("date_of_update", Date),

    Column("latitude", Float),
    Column("longitude", Float),
    Column("location", Text),

    Column("month", Integer),
    Column("day_of_week", String(12))
)

# Confirm that the table schema matches every column
# in the cleaned, feature-engineered DataFrame.
assert set(db_df.columns) == set(
    crime_table.columns.keys()
)

# Create the crime table.
# This does not delete existing database records.
metadata.create_all(engine)


# ============================================================
# STEP 5: VALIDATE THE DATABASE SCHEMA
# ============================================================

# Inspect the actual SQLite table definition.
db_inspector = inspect(engine)

schema_columns = db_inspector.get_columns(
    "crime_records"
)

schema_summary = pd.DataFrame([
    {
        "column_name": column["name"],
        "data_type": str(column["type"]),
        "primary_key": bool(column["primary_key"])
    }
    for column in schema_columns
])

# Ensure the intended primary key exists.
primary_key_info = db_inspector.get_pk_constraint(
    "crime_records"
)

assert primary_key_info["constrained_columns"] == [
    "case_number"
]

print("\n========== CRIME TABLE SCHEMA ==========")
display(schema_summary)


# ============================================================
# STEP 6: INSERT CLEANED CRIME DATA
# ============================================================

# Use one database transaction for insertion.
with engine.begin() as connection:

    existing_count = connection.execute(
        text("SELECT COUNT(*) FROM crime_records")
    ).scalar_one()

    if existing_count == 0:

        # Append records to the SQLAlchemy-created table.
        # The database schema is not replaced.
        db_df.to_sql(
            name="crime_records",
            con=connection,
            if_exists="append",
            index=False,
            chunksize=200
        )

        print("\nCrime data inserted successfully.")

    else:

        # Avoid duplicate insertion if this cell is rerun.
        existing_cases = pd.read_sql_query(
            text("SELECT case_number FROM crime_records"),
            connection
        )

        assert existing_count == len(db_df), (
            "Existing database record count differs "
            "from the cleaned dataset."
        )

        assert set(existing_cases["case_number"]) == set(
            db_df["case_number"]
        ), (
            "Existing database records differ "
            "from the cleaned dataset."
        )

        print("\nCrime data already exists. No duplicates added.")


# ============================================================
# STEP 7: VERIFY INSERTED DATABASE RECORDS
# ============================================================

with engine.connect() as connection:

    # Count records stored in the crime table.
    database_count = connection.execute(
        text("SELECT COUNT(*) FROM crime_records")
    ).scalar_one()

    # Count unique case numbers.
    unique_cases = connection.execute(
        text("""
            SELECT COUNT(DISTINCT case_number)
            FROM crime_records
        """)
    ).scalar_one()

    # Read a sample of stored crime records.
    database_sample = pd.read_sql_query(
        text("""
            SELECT
                id,
                case_number,
                date,
                iucr_code,
                arrest,
                district_code,
                year,
                month,
                day_of_week
            FROM crime_records
            ORDER BY id
            LIMIT 5
        """),
        connection
    )

    # Check SQLite database integrity.
    integrity_result = connection.execute(
        text("PRAGMA integrity_check")
    ).scalar_one()


# ============================================================
# STEP 8: DISPLAY THE DATABASE RESULTS
# ============================================================

print("\n========== DATABASE RECORD VALIDATION ==========")

print("Cleaned Crime Records:", len(db_df))
print("Database Crime Records:", database_count)
print("Unique Case Numbers:", unique_cases)
print("Database Columns:", len(schema_columns))

print("\n========== SAMPLE DATABASE RECORDS ==========")
display(database_sample)


# ============================================================
# STEP 9: VALIDATE THE DATABASE
# ============================================================

# Every cleaned record must be stored in the database.
assert database_count == len(db_df) == 498

# The database must contain 498 unique case numbers.
assert unique_cases == 498

# Verify that all 22 columns were created.
assert len(schema_columns) == 22

# Verify SQLite integrity.
assert integrity_result == "ok"

# Verify the database file was generated.
assert DB_FILE.is_file()


# ============================================================
# STEP 10: FINAL TASK SUMMARY
# ============================================================

print("\n========== USE CASE 1 - TASK 5 SUMMARY ==========")

print("Database:", DB_FILE.name)
print("Table Created: crime_records")
print("Primary Key: case_number")
print("Total Columns:", len(schema_columns))
print("Total Records Inserted/Verified:", database_count)
print("SQLite Integrity:", integrity_result)

print("\nUSE CASE 1 - TASK 5: PASS")
```

### Assessment Questions and Answers

Identifies unique crime types by joining crime records with the IUCR reference dataset and calculating category frequencies. Examines date formats, year consistency, chronological anomalies and yearly crime distribution while highlighting uneven sample coverage and unavailable occurrence times.

```python

# ============================================================
# USE CASE 1 - CODE CELL 6
# ASSESSMENT QUESTIONS AND ANSWERS
# ============================================================

# Assessment Questions:
# Q1. How many unique crime types exist in the dataset?
# Q2. Are there anomalies in date formats or obvious outliers?
#
# Methods:
# pd.read_csv() reads the IUCR reference dataset.
# merge() links crime records with Primary Type.
# nunique() calculates the number of crime types.
# value_counts() calculates category and year counts.
# pd.to_datetime() validates original crime dates.
# dt.year extracts years for anomaly checking.
#
# Important:
# Primary Type is obtained from the IUCR reference file.
# An IUCR code is not the same as a crime category.
# No original crime records are removed or modified.


# ============================================================
# STEP 1: IMPORT REQUIRED LIBRARIES
# ============================================================

import pandas as pd
from pathlib import Path
from IPython.display import display


# ============================================================
# STEP 2: VALIDATE PREVIOUS DATAFRAMES
# ============================================================

# crime_df was created in Code Cell 1.
# feature_df was created in Code Cell 3.
assert "crime_df" in globals(), (
    "Run Code Cell 1 before this cell."
)

assert "feature_df" in globals(), (
    "Run Code Cell 3 before this cell."
)

assert len(crime_df) == 498
assert len(feature_df) == 498


# ============================================================
# QUESTION 1: HOW MANY UNIQUE CRIME TYPES EXIST?
# ============================================================

# STEP 3: Locate the original IUCR reference file.
IUCR_FILE = (
    BASE_DIR / "data" / "raw" /
    "iucr_codes.csv.csv"
)

# Support the standard filename if it was corrected.
if not IUCR_FILE.is_file():
    IUCR_FILE = (
        BASE_DIR / "data" / "raw" /
        "iucr_codes.csv"
    )

assert IUCR_FILE.is_file(), (
    f"IUCR reference dataset not found: {IUCR_FILE}"
)


# STEP 4: Load the IUCR reference dataset.
iucr_df = pd.read_csv(
    IUCR_FILE,
    dtype=str,
    low_memory=False
)

# Standardize reference column names.
# Example: "PRIMARY_TYPE " becomes "primary_type".
iucr_df.columns = (
    iucr_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# Match the IUCR reference key to the master dataset.
iucr_df = iucr_df.rename(
    columns={"iucr": "iucr_code"}
)

assert {
    "iucr_code",
    "primary_type"
}.issubset(iucr_df.columns)


# STEP 5: Standardize IUCR codes before joining.
# Example: 486 becomes 0486.
crime_codes = feature_df[
    ["iucr_code"]
].copy()

crime_codes["iucr_code"] = (
    crime_codes["iucr_code"]
    .astype("string")
    .str.strip()
    .str.upper()
    .str.zfill(4)
)

iucr_df["iucr_code"] = (
    iucr_df["iucr_code"]
    .astype("string")
    .str.strip()
    .str.upper()
    .str.zfill(4)
)

iucr_df["primary_type"] = (
    iucr_df["primary_type"]
    .astype("string")
    .str.strip()
    .str.upper()
)

# Each IUCR code should identify one reference record.
assert iucr_df["iucr_code"].is_unique


# STEP 6: Join crime records with IUCR Primary Type.
crime_with_types = crime_codes.merge(
    iucr_df[
        ["iucr_code", "primary_type"]
    ],
    on="iucr_code",
    how="left",
    validate="many_to_one"
)

# Ensure every source crime was matched.
assert len(crime_with_types) == len(feature_df)

assert crime_with_types["primary_type"].notna().all(), (
    "Some crime IUCR codes could not be matched."
)


# STEP 7: Calculate unique crime types.
unique_crime_types = (
    crime_with_types["primary_type"].nunique()
)

# Calculate frequency of each Primary Type.
crime_type_counts = (
    crime_with_types["primary_type"]
    .value_counts()
    .rename_axis("primary_type")
    .reset_index(name="crime_count")
)

print("========== QUESTION 1 ==========")

print(
    "Question: How many unique crime types "
    "exist in the dataset?"
)

print(
    "Answer:",
    unique_crime_types,
    "unique crime types."
)

print("\n========== CRIME TYPE FREQUENCIES ==========")

display(crime_type_counts)

# Validate the classification results.
assert unique_crime_types == 23

assert crime_type_counts["crime_count"].sum() == 498


# ============================================================
# QUESTION 2: ARE THERE DATE ANOMALIES OR OUTLIERS?
# ============================================================

# STEP 8: Read the original date strings.
# We use crime_df because it preserves original CSV values.
raw_dates = (
    crime_df["date"]
    .astype("string")
    .str.strip()
)


# STEP 9: Validate original date formats.
# The supplied source uses month/day/year.
parsed_dates = pd.to_datetime(
    raw_dates,
    format="%m/%d/%Y",
    errors="coerce"
)

# Invalid or missing dates become NaT.
invalid_date_count = int(
    parsed_dates.isna().sum()
)

# Check whether actual occurrence times were supplied.
records_with_time = int(
    raw_dates.str.contains(
        r"\b\d{1,2}:\d{2}(?::\d{2})?\b",
        regex=True,
        na=False
    ).sum()
)


# STEP 10: Validate the year range.
# The assessment describes crime data from 2012 to 2016.
crime_years = parsed_dates.dt.year

year_outliers = (
    (crime_years < 2012)
    |
    (crime_years > 2016)
)

year_outlier_count = int(
    year_outliers.sum()
)

# Check consistency between the Date and Year fields.
original_year = pd.to_numeric(
    crime_df["year"],
    errors="coerce"
)

year_mismatch_count = int(
    (
        original_year.ne(crime_years)
        |
        original_year.isna()
    ).sum()
)


# STEP 11: Check the record-update dates.
update_dates = pd.to_datetime(
    crime_df["date_of_update"],
    format="%m/%d/%Y",
    errors="coerce"
)

invalid_update_dates = int(
    update_dates.isna().sum()
)

# An update occurring before the incident would be
# a potential chronology anomaly.
updates_before_crime = int(
    (update_dates < parsed_dates).sum()
)


# STEP 12: Display crime distribution by year.
yearly_distribution = (
    crime_years
    .value_counts()
    .sort_index()
    .rename_axis("year")
    .reset_index(name="crime_count")
)

print("\n========== CRIME DISTRIBUTION BY YEAR ==========")

display(yearly_distribution)


# STEP 13: Calculate sample concentration.
records_in_2016 = int(
    (crime_years == 2016).sum()
)

percentage_in_2016 = (
    records_in_2016 / len(crime_df) * 100
)

print("\n========== DATE ANOMALY VALIDATION ==========")

print("Total Crime Records:", len(crime_df))
print("Invalid Crime Dates:", invalid_date_count)
print("Years Outside 2012-2016:", year_outlier_count)
print("Date and Year Mismatches:", year_mismatch_count)
print("Invalid Update Dates:", invalid_update_dates)
print("Updates Before Crime Date:", updates_before_crime)

print("Records With Occurrence Time:", records_with_time)

print("Earliest Crime Date:", parsed_dates.min().date())
print("Latest Crime Date:", parsed_dates.max().date())

print("Crimes Recorded in 2016:", records_in_2016)
print(
    "Percentage Recorded in 2016:",
    round(percentage_in_2016, 2),
    "%"
)


# STEP 14: Answer the second assessment question.
print("\n========== QUESTION 2 ==========")

print(
    "Question: Are there anomalies in date "
    "formats or obvious outliers?"
)

if (
    invalid_date_count == 0
    and year_outlier_count == 0
    and year_mismatch_count == 0
    and invalid_update_dates == 0
    and updates_before_crime == 0
):

    print(
        "Answer: No invalid date formats, "
        "out-of-range years, or chronological "
        "inconsistencies were detected."
    )

else:

    print(
        "Answer: Potential date anomalies were detected. "
        "Review the validation results above."
    )

print(
    "Observation: Crime records are heavily "
    "concentrated in 2016."
)

print(
    "This indicates uneven sample coverage, "
    "not necessarily incorrect data."
)

if records_with_time == 0:
    print(
        "Limitation: Crime occurrence times are "
        "not available in the supplied CSV."
    )


# ============================================================
# STEP 15: FINAL VALIDATION
# ============================================================

assert len(crime_with_types) == 498
assert unique_crime_types == 23

assert invalid_date_count == 0
assert year_outlier_count == 0
assert year_mismatch_count == 0
assert invalid_update_dates == 0
assert updates_before_crime == 0

# The original DataFrames have not been modified.
assert len(crime_df) == 498
assert len(feature_df) == 498

print("\nUSE CASE 1 - ASSESSMENT QUESTIONS: PASS")
```
